## Step 3: Train the Model
Now we train the YOLOv8 model using your unzipped dataset.

```python
from ultralytics import YOLO
import yaml

# Setting the correct absolute path to handle the double-folder from the zip file
yaml_path = '/content/yolo_dataset/yolo_dataset/dataset.yaml'

with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

# Set base path for Colab
data['path'] = '/content/yolo_dataset/yolo_dataset'

with open(yaml_path, 'w') as f:
    yaml.dump(data, f)

# Load a YOLOv8 Nano model (use yolov8n-seg.pt if your data has segmentation masks, or yolov8n.pt for boxes)
model = YOLO('yolov8n.pt')

# Train the model for 50 epochs
results = model.train(data=yaml_path, epochs=50, imgsz=640)
```
