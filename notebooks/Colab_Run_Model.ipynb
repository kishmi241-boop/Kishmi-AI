## Step 3: Run the AI Scanner on the Image
This will take the `best.pt` model weights and scan the sample image.

```python
from ultralytics import YOLO
import glob
import os

# Find where best.pt is located inside the unzipped model folder
model_paths = glob.glob('/content/model_files/**/best.pt', recursive=True)
if not model_paths:
    print("Could not find best.pt. Make sure trained_model.zip is uploaded correctly.")
else:
    best_model_path = model_paths[0]
    print(f"Found model at: {best_model_path}")
    
    # Load the trained model
    model = YOLO(best_model_path)
    
    # Run prediction on the sample image
    print("Scanning image...")
    results = model.predict(source='/content/test_images/', save=True, imgsz=640, conf=0.25)
    print("\nScanning complete!")
```

## Step 4: View the Scanned Image
Run this to display the image with the AI's detected skin conditions drawn on it.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import glob

# Find the latest predict folder
predict_folders = sorted(glob.glob('/content/runs/detect/predict*'))
if predict_folders:
    latest_folder = predict_folders[-1]
    images = glob.glob(f'{latest_folder}/*.jpg')
    
    # Display up to 5 images
    for img_path in images[:5]:
        img = mpimg.imread(img_path)
        plt.figure(figsize=(10,10))
        plt.imshow(img)
        plt.axis('off')
        plt.show()
else:
    print("No predictions found.")
```
